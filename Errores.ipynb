```python
usuarios = {
    "ANA": "1234",
    "JUAN": "abcd"
}

try:
    usuario = input("Usuario: ").upper()
    password = input("Contraseña: ")

    if usuario not in usuarios:
        raise KeyError("El usuario no existe")

    if usuarios[usuario] != password:
        raise ValueError("Contraseña incorrecta")

    else:
        print("Bienvenido,", usuario)

except KeyError as e:
    print("Error:", e)

except ValueError as e:
    print("Error:", e)
```

```python
productos = {
    "pan": 1.5,
    "leche": 2.0,
    "huevos": 3.0
}

print("Productos disponibles:", productos)

try:
  nombre = input("¿Que producto quieres?: ").lower()
  if nombre not in productos:
    raise KeyError("Producto no disponible")

  cantidad = int(input("Ingrese la cantidad: "))

  if cantidad <= 0:
    raise ValueError("La cantidad debe ser mayor a 0")

  total = productos[nombre] * cantidad
  print("Total a pagar:", total)

except KeyError as e :
  print("Producto no disponible")
except ValueError as e:
  print("La cantidad debe ser un número entero", e)
```

INTRODUCCIÓN A LOS LENGUAJES REGULARES


Una expresion regular es un secuencia de caracteres que define un patrón de busqueda
Se utilizan para :


*  Extraer datos
*  Reemplazar palabras
*  Limpiar cadenas de textos

En Python se trabaja con el modulo re








```python
import re
```

```python
import re
texto  = "Python es un lenguaje"
resultado = re.search("Python", texto)
print(resultado)
```

```python
resultado1 = re.search("x", texto)
print(resultado1)
```

```python
resultado1 = re.search("a", texto)
print(resultado1)
```

El método search permite encontrar los índices donde por primera vez se encuentra una cadena específica.

```python
import re
texto = "Tengo 25 perros y 4 gatos"
resultado = re.findall(r"\d", texto)
print(resultado)
```

El método findall permite encontrar números (\d significa dígitos). No determina cadenas de solo dígitos.

```python
import re
texto = "Tengo 23 perros y 3 gatos"
resultado = re.findall(r"\d+", texto)
print(resultado)
```

```python
import re
texto = "Python es divertido"
resultado = re.findall(r"\w+", texto)
print(resultado)
```

Tambiéb encontrar palabras completas (mediante \w+")

```python
import re
texto = "Universidad"
resultado = re.findall(r"[aeiou]", texto)
print(resultado)
```

También encuentra letras específicas.

```python
import re
texto = "Python Es Genial"
resultado = re.findall(r"[A-Z]", texto)
print(resultado)
```

Se pueden establecer letras mayúsculas.

```python
import re
texto = "Hola mundo Python"
resultado = re.findall(r"\s", texto)
print(resultado)
```

El argumento \s representa espacios.

```python
import re
texto = "Tengo 15 libros y 200 hojas"
resultado = re.findall(r"\d+", texto)
print(resultado)
```

\d+ encuentra números completos.

```python
import re
texto = "Me gusta Java"
resultado = re.sub("Java","Python", texto)
print(resultado)
```

Se puede reemplazar cadenas por otras

```python
import re
texto = "rojo,verde,azul"
nuevo = re.split(",",texto)
print(nuevo)
```

Se puede separar texto

```python
import re
texto = "Correos: ana@gmail.com y juan@hotmail.com"
resultado = re.findall(r"\w+@\w+\.com", texto)
print(resultado)
```

Hacer una búsqueda sencilla de correos electrónicos.

```python
import re
telefono = "3104567890"
resultado = re.fullmatch(r"\d{10}", telefono)
print(resultado)
```

fullmatch valida toda la cadena. En este caso que se tengan 10 dígitos.

```python
import re
texto = "color colour"
resultado = re.findall(r"colou?r", texto)
print(resultado)
```

La letra u puede aparecer o puede no aparecer.

Resumen rápido

\d → dígitos

\w → letras y números

\s → espacios

. → cualquier carácter

 + → una o más veces

 * → cero o más veces

? → opcional

[abc] → a, b o c

[^abc] → excepto a, b o c
