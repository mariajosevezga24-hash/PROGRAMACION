CONVERTIR DESDE LISTA

```python
lista = ["a", "b", "c", "d"]

tupla_desde_lista = tuple(lista)
diccionario_desde_lista = dict(enumerate(lista))
conjunto_desde_lista = set(lista)

print("Tupla:", tupla_desde_lista)
print("Diccionario:", diccionario_desde_lista)
print("Conjunto:", conjunto_desde_lista)
```

CONVERTIR DESDE TUPLA

```python
tupla = ("Maria", "Juan", "Jose")

lista_desde_tupla = list(tupla)
diccionario_desde_tupla = dict(enumerate(tupla))
conjunto_desde_tupla = set(tupla)

print("Lista:", lista_desde_tupla)
print("Diccionario:", diccionario_desde_tupla)
print("Conjunto:", conjunto_desde_tupla)
```

CONVERTIR DESDE DICCIONARIO

```python
diccionario = {"a": 1, "b": 2, "c": 3}

lista_desde_dic = list(diccionario.items())
tupla_desde_dic = tuple(diccionario.items())
conjunto_desde_dic = set(diccionario.items())

print("Lista:", lista_desde_dic)
print("Tupla:", tupla_desde_dic)
print("Conjunto:", conjunto_desde_dic)
```

CONVERTIR DESDE CONJUNTO

```python
conjunto = {10, 20, 30}

lista_desde_conj = list(conjunto)
tupla_desde_conj = tuple(conjunto)
diccionario_desde_conj = dict(enumerate(conjunto))

print("Lista:", lista_desde_conj)
print("Tupla:", tupla_desde_conj)
print("Diccionario:", diccionario_desde_conj)
```
