CONJUNTOS

```python

numeros = {1, 2, 3, 4, 4, 2}
print("Conjunto (sin repetidos):", numeros)

numeros.add(5)
print("Después de agregar:", numeros)

numeros.remove(2)
print("Después de eliminar:", numeros)

```

```python
# CONJUNTOS Y OPERACIONES

# Conjunto base
conjunto = {1, 2, 3, 4}

# Crear conjuntos desde listas (elimina repetidos)
conjunto_2 = set([5, 3, 5, 6])
conjunto_3 = set([4, 2])

# Mostrar conjuntos
print("Conjunto 1:", conjunto)
print("Conjunto 2:", conjunto_2)
print("Conjunto 3:", conjunto_3)

# Intersección (elementos en común)
print("Intersección (método):", conjunto.intersection(conjunto_2))
print("Intersección (&):", conjunto & conjunto_2)

# Unión (todos los elementos sin repetir)
print("Unión:", conjunto | conjunto_2)

# Diferencia (elementos que están en conjunto pero no en conjunto_2)
print("Diferencia:", conjunto - conjunto_2)

# Intersección con conjunto_3
print("Intersección con conjunto_3:", conjunto & conjunto_3)
```
