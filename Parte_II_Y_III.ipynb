```python

import numpy as np
import pandas as pd

np.random.seed(123)

grupo_A = np.random.normal(loc=7.2, scale=0.4, size=54)
grupo_B = np.random.gamma(shape=306, scale=0.0229, size=54)

df = pd.DataFrame({
    "A": grupo_A,
    "B": grupo_B
})
```

```python
df.describe()
```

```python
from scipy.stats import shapiro # Prueba shapiro para la normal
shapiro(df["A"])
```

```python
from scipy.stats import shapiro # Prueba shapiro para la Gamma
shapiro(df["B"])
```

No se rechaza la hpótesis de normalidad con una confianza del 95%

```python
from scipy import stats
t_stat, p_val = stats.ttest_ind(df["A"], df["B"])

print("Estadístico t:", t_stat)
print("Valor p:", p_val)
```

No se rechaza con una confianza del 95% que los puntajes de ambas encuestas estan es igual

```python
stats.kendalltau(df["A"], df["B"])
```

No se rechaza con una confianza del 95% que los puntajes de ambas encuestas no están correlacionados pues p>0.05

```python
stats.spearmanr(df["A"], df["B"])
```
