```python
def contador_vocales1(palabra):
  vocales = "aeiou"
  letras1 = "bcdfghjklmnñpqrstvwxyz"
  vocales2 =  vocales.upper()
  letra2 = letras1.upper()
  numero = "0123456789"
  espacio = " "

  contador = 0
  contador1 = 0
  contador2 = 0
  contador3 = 0
  contador4 = 0
  contador5 = 0

  for letra in palabra:
    if letra in vocales:
      contador += 1
    elif letra in vocales2:
      contador1 += 1
    if letra in letras1:
      contador2 += 1
    elif letra in letra2:
      contador3 += 1
    if letra in numero:
      contador4 += 1
    else:
      contador5 += 1

  return({"Cantidad de vocales:": contador,
          "Cantidad de Vocales Mayusculas:": contador1,
          "Cantidad de Consonantes:": contador2,
          "Cantidad de Consonantes Mayusculas:": contador3,
          "Cantidad de Números:": contador4,
          "Cantidad de otros caracteres:": contador5})
```
